```python

import cv2
import matplotlib.pyplot as plt  
 
# 读入原图片
#Read in original image
img = cv2.imread('yahboom.jpg')
# 打印出图片尺寸
#Print out image size
print(img.shape)
# 将图片高和宽分别赋值给x，y
#Assign the height and width of the image to x and y respectively
x, y = img.shape[0:2]
 
# 缩放到原来的二分之一，输出尺寸格式为（宽，高）
#Zoom to half of the original size, output size format is (width, height)
img_test1 = cv2.resize(img, (int(y / 2), int(x / 2)))
# cv2.imshow('resize0', img_test1)
# cv2.waitKey()
 
# 最近邻插值法缩放
# 缩放到原来的四分之一
#Nearest Neighbor Interpolation Scaling
#Zoom to the original quarter
img_test2 = cv2.resize(img, (0, 0), fx=0.25, fy=0.25, interpolation=cv2.INTER_NEAREST)
# cv.imshow('resize1', img_test2)
# cv.waitKey()
# cv.destroyAllWindows()
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
dst1 = cv2.cvtColor(img_test1, cv2.COLOR_BGR2RGB)
dst2 = cv2.cvtColor(img_test2, cv2.COLOR_BGR2RGB)

# 显示原始图像
#Show original image
plt.imshow(img)
plt.show()
```

```python
# 显示缩放1/2
#Display Zoom 1/2
plt.imshow(dst1)
plt.show()
```

```python
# 显示缩放1/4 邻插值法缩放
#Display Zoom 1/4 Neighbor Interpolation Zoom
plt.imshow(dst2)
plt.show()
```

```python
import numpy as np 
from matplotlib import pyplot as plt 

x = np.arange(1,11) 
y =  2 * x +  5 
plt.title("Matplotlib demo") 
plt.xlabel("x axis caption") 
plt.ylabel("y axis caption") 
plt.plot(x,y) 
plt.show()
```
